```python
summary = pd.DataFrame({
    "Data Type": df.dtypes,
    "Missing Values": df.isnull().sum(),
    "Missing %": (df.isnull().sum() / len(df) * 100).round(2),
    "Unique Values": df.nunique()
})

summary.sort_values("Missing %", ascending=False)
```

```python
df.isnull().sum()
```

```python
missing_percentage = (df.isnull().sum() / len(df) * 100).sort_values(ascending=False)

missing_percentage
```

```python
df.isnull().any().sum()
```

```python
missing_percentage[missing_percentage > 0].plot(
    kind="barh",
    figsize=(15, 8),
    color="darkblue"
)

plt.title("Total chart of Missing Values",fontsize=15, fontweight="bold", color="black")
plt.xlabel("Missing Percentage (%)",fontsize=14, fontweight="bold", color="black")
plt.ylabel("Column Names",fontsize=14, fontweight="bold", color="black")
plt.xticks(fontsize=10, fontweight="bold", color="black")
plt.yticks(fontsize=10, fontweight="bold", color="black")
plt.tight_layout()
plt.grid()
plt.show()

```
